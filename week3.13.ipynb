BREAK, CONTINUE AND PASS

```python
email = input()
print("E-mail:",email)
for c in email:
    if(c=='@'):
        break #stops instantly
    print(c, end='')
```

```python
email = input()
print("E-mail:",email)
for c in email:
    if(c=='@'):
        print('')#used to send the another half to next line
        continue #skips that specific one
    print(c, end='')
    #end is used to write in sinle line instead of veriaclly in diff-diff lines
```
