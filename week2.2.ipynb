A programmer's perspective

```python
ram_bank_balance=1000000
#bank balance of ram
ram_loan=345600
#loan of ram
lakshman_bank_balance=1000000
#bank balance of lakshman
lakshman_loan=902000
#loan of lakshman
net_income=ram_bank_balance+lakshman_bank_balance
#net income is sum of all source of income's amt
net_liability=ram_loan+lakshman_loan
#liability is the sum of all type of loan
final_value=net_income-net_liability
#final value mean final amt, it shows the actual your values
print(final_value)
```
