```python
#if condition
#Let us consider the movie "Avengers".This is a 13+ movie.
birth_year=int(input("Please enter your DOB. "))
current_year=2021
age=current_year-birth_year
if(age<13):
    print("You are not allowed to watch this movie.")
else:
    print("You are allowed to watch this movie. ")

print("Have a nice time")
```

```python
#check the number whether it is odd are even
n=int(input("Enter a number:"))
if(n%2==0):
    print("Entered number is even")
else:
    print("Entered number is odd")

```

```python
#Find whether the number ends with 0 or 5 or any other number
n=int(input("Enter any number"))
if(n%10==0):
    print("Entered number ends with 0")
elif((n%5==0)and(n%10!=0)):
    print("Entered number ends with 5")
else:
    print("Entered number ends with other than nuber 0 & 5")
```

```python
#Find the grade of student based on the given marks from 0 to 100
m=int(input("Enter your marks to see your grades:"))
if(m>=90):
    print("Your grade is A")
elif(m>=80):
    print("Your grade is B")
elif(m>=70):
    print("Your grade is C")
elif(m>=60):
    print("Your grade is D")
else:
    print("Fail!!")
```
