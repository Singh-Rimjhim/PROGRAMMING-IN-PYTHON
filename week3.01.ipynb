```python
#Factorial of a given number
print("Enter number")
n=int(input())
i=1
answer=1
while(i<=n):
    answer=(i*answer)
    i=i+1

print(answer)
    
```

```python
#Reverse the digits in the given number
num = int(input("Enter a number: "))
absNum = abs(num)
rev = absNum%10
absNum =absNum//10
while(absNum>0):
    r= absNum % 10
    absNum = absNum//10
    rev = rev*10 +absNum%10
if(num<0):
    rev = rev - 2*rev
print(rev)
```

```python
#Find whether the entered number is palindrome or not
num = int(input("Enter a number: "))
print(num)
absNum = abs(num)
rev = absNum%10
absNum =absNum//10
while(absNum>0):
    r= absNum % 10
    absNum = absNum//10
    rev = rev*10 + r
if(num<0):
    rev = rev - 2*rev
if(rev==num):
    print("Palindrome")
else:
    print("Not a Plindrome!!")
```
