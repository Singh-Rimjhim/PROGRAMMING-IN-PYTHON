Escape characters and type of quotes

What is escape characters?

like we want to write(it's good) in single quotes i.e,'it's a good day.' then 'it' will be in single quote and rest will be as it is , then here comes the concept of escape characters
 

```python
#Evample:
print('It's a good day')
```

Insert a blackslash before 's, it will be the use of escape character

```python
print('It\'s a beautiful day')
print("We are from \"IIT\" Madras")
```

```python
print('My name is Jhalak. I am from home')
#it will print it but what if we want more space b/w them
```

Use \t for more space between words

```python
print('My name is Jhalak.\t I am from home')
```

Use \n for new line

```python
print('My name is Jhalak.\n I am from home')
```
