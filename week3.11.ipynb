```python
num=int(input())
if (num > 2):
    print(2, end=' ')
for i in range(3, num):
    flag = False
    for j in range(2, i):
        if (i%j==0):
            flag=False
            break
        else:
            flag = True
    if (flag):
        print(i, end=' ')
    
```

```python
empID = input('Enter Employee ID: ')
while(empID!='-1'):
    trade = int(input('Enter trade amount: '))
    profit_loss = 0
    while(trade != 0):
        profit_loss = profit_loss + trade
        trade = int(input('Enter trade amount: '))
        print(f'Profit/Loss for employee {empID} is {profit_loss}')
        empID = input("\nEnter Employee ID:")
```
