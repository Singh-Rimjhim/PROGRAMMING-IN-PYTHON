STRING METHODS

```python
x= 'string methods'
```

lower():- Converts a string into lower case

```python
print(x.lower())
```

upper():-Converts a string into lower case

```python
print(x.upper())
```

capitalize():-Converts the first character to upper case 

```python
print(x.capitalize())
```

title():-Converts the first character of each word to upper case

```python
print(x.title())
```

swapcase():-Swaps cases, lower case becomes upper case and vice versa

```python
print(x.swapcase())
```

islower():-Returns True if all characters in string are lower case 

```python
x='python'
print(x.islower())
x='Python'
print(x.islower())
```

isupper():-Returns True if all characters in string are upper case 

```python
x='PYTHON'
print(x.isupper())
x='Python'
print(x.isupper())
```

istitle():-Returns True if the string follows the rule of a title

```python
x= 'string Methods'
print(x.istitle())
```

isdigit():-Returns true if all characters in the string are digit

```python
x='100'
print(x.isdigit())
x='100abc'
print(x.isdigit())
```

isalpha():-Returns TRUE if all characters in the string are in alphabets

```python
x='abc'
print(x.isalpha())
x='100abc'
print(x.isalpha())
```

isalnum():-Returns TRUE if all the characters are alpha numeric

```python
x='100abc#@'
print(x.isalnum())
x='100abc'
print(x.isalnum())
```

strip():-Returns a trimmed verison of string

```python
x='------------------Python-----------------------'
print(x.strip('-'))
```

lstrip():-Returns a left trimmed verison of string

```python
x='------------------Python-----------------------'
print(x.lstrip('-'))
```

rstrip():-Returns a right trimmed verison of string

```python
x='------------------Python-----------------------'
print(x.rstrip('-'))
```

stratwith():-Returns True if the string strats with specified value

```python
x='Python'
print(x.startswith('P'))
print(x.startswith('p'))
```

endwith():-Returns True if the string endss with specified value

```python
x='Python'
print(x.endswith('n'))
print(x.endswith('N'))
```

count():-Returns the number of times a specified value occurs in a string

```python
x='Jhalak loves ritu'
print(x.count('t'))
print(x.count('l'))
```

index():-Searches the string for a specified value and returns the position of where it was found

```python
print(x.index('s'))
print(x.index('l'))
```

replace():-Returns a string where a specified value is replaced with specified value

```python
print(x.replace('r','R'))
```
