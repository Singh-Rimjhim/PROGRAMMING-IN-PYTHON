FORMATTED PRINTING

```python
for x in range (10):
    print(x, end = ' ')
```

```python
d=10
m=5
y=2021
#print("Today's date is ", d, m, y, sep = '/')
#output: Today's date is /10/5/2021     
# sep = '/' will use to seperate write it as the actual date and it will present the every variable written in the same print line
#but / is also in front of 10 so another way to write is :
print("Today's date is", end = ' ')
print(d, m, y, sep='/')
#any other thing instead of / can be used anything like -
```

```python
num = int(input())
for i in range(1, 11):
    print(num, 'X', i, '=', num*i)
    print(f'{num} X {i}={num*i}')
```

```python
pi = 22/7
print(f'Value of PI ={pi}')
print('Value pf PI = {0}'.format(pi))
print('Value of PI = %f' % (pi))
```

```python
print('{0}'.format(1))
print('{0}'.format(11))
print('{0}'.format(111))
print('{0}'.format(1111))
print('{0}'.format(11111))
```

```python
print('{0:5d}'.format(1))
print('{0:5d}'.format(11))
print('{0:5d}'.format(111))
print('{0:5d}'.format(1111))
print('{0:5d}'.format(11111))
```
